```python
def read_json_gz(path):
    # Most datasets on McAuley's page are one JSON object per line.
    # Some older ones need eval() instead of json.loads().
    with gzip.open(path, 'rt') as f:
        for line in f:
            yield json.loads(line)

data = list(read_json_gz(DATA_PATH))
df = pd.DataFrame(data)
print(len(df), "rows")
df.head()
```
